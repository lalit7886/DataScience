```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
x=2*np.random.rand(100,1)
y= 4*x + 3 + np.random.randn(100,1)
```

```python
plt.scatter(x,y)
plt.xlabel("X")
plt.ylabel("y")
```

```python
class RidgeRegression:
    def __init__(self,regularization_parameter=0.5,learning_rate=0.001,epochs=1000):
        self.regularization_parameter=regularization_parameter
        self.learning_rate=learning_rate
        self.epochs=epochs
        self.weight=None
        self.bias=None
        
    
    def predict(self,x):
        return self.weight*x + self.bias
    
    def compute_cost(self,y,y_pred):
        cost=0.5*np.mean((y-y_pred)**2) + self.regularization_parameter * np.sum(self.weight**2)
        return cost
        
    def fit(self,x,y):
        n=x.shape[0]
        self.weight=0
        self.bias=0
        for epochs in range(self.epochs):
            y_pred=self.predict(x)
            
            dw=(-1/n)*np.sum(x*(y-y_pred)) + 2*self.regularization_parameter*np.sum(self.weight)
            db=(-1/n)*np.sum(y-y_pred)
            
            self.weight -= self.learning_rate*dw
            self.bias -= self.learning_rate*db


```

```python
model = RidgeRegression(
    learning_rate=0.1,
    epochs=1000
)

model.fit(x, y)

print("Weight:", model.weight)
print("Bias:", model.bias)
```
